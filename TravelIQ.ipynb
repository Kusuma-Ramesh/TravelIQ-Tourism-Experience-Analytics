TravelIQ is a tourism analytics and machine learning platform developed to analyze traveler behavior, attraction experiences, and tourism patterns using historical tourism data.

````python
import pandas as pd
import os

excel_dfs = {}
tourism_dataset_base_path = '/content/Tourism Dataset'

print("Loading Excel files and displaying their heads:")
for root, _, files in os.walk(tourism_dataset_base_path):
    for file_name in files:
        if file_name.endswith('.xlsx'):
            file_path = os.path.join(root, file_name)

            df_name = os.path.splitext(file_name)[0].replace(' ', '_').replace('.', '_').lower() + '_df'
            try:
                df_loaded = pd.read_excel(file_path)
                excel_dfs[df_name] = df_loaded
                print(f"\n--- {file_name} ({df_name}) ---")
                print(df_loaded.head())
            except Exception as e:
                print(f"Error loading {file_path}: {e}")

print("\nAll Excel files loaded into 'excel_dfs' dictionary.")
print("Available DataFrames:", excel_dfs.keys())
````

````python
import pandas as pd
import numpy as np

df = excel_dfs['transaction_df'].copy()

user_df_processed = excel_dfs['user_df'].copy()
user_df_processed = user_df_processed.merge(excel_dfs['city_df'][['CityId', 'CityName']].rename(columns={'CityName': 'Customer_City'}), on='CityId', how='left')
user_df_processed = user_df_processed.merge(excel_dfs['country_df'][['CountryId', 'Country']].rename(columns={'Country': 'Customer_Country'}), on='CountryId', how='left')
user_df_processed = user_df_processed.merge(excel_dfs['region_df'][['RegionId', 'Region']].rename(columns={'Region': 'Customer_Region'}), on='RegionId', how='left')
user_df_processed = user_df_processed.merge(excel_dfs['continent_df'][['ContinentId', 'Continent']].rename(columns={'Continent': 'Customer_Continent'}), on='ContinentId', how='left')

user_cols_to_merge = ['UserId', 'Customer_City', 'Customer_Country', 'Customer_Region', 'Customer_Continent']
df = df.merge(user_df_processed[user_cols_to_merge], on='UserId', how='left')

item_df_processed = excel_dfs['item_df'].copy()
item_df_processed = item_df_processed.merge(excel_dfs['city_df'][['CityId', 'CityName']].rename(columns={'CityId': 'AttractionCityId', 'CityName': 'AttractionCityName'}), on='AttractionCityId', how='left')
item_df_processed = item_df_processed.merge(excel_dfs['type_df'], on='AttractionTypeId', how='left')

item_cols_to_merge = ['AttractionId', 'Attraction', 'AttractionAddress', 'AttractionCityName', 'AttractionType']
df = df.merge(item_df_processed[item_cols_to_merge], on='AttractionId', how='left')

df = df.rename(columns={'VisitMode': 'VisitModeId_transaction'})
df = df.merge(excel_dfs['mode_df'], left_on='VisitModeId_transaction', right_on='VisitModeId', how='left')
df = df.rename(columns={'VisitMode_y': 'VisitModeName'})
df.drop(columns=['VisitModeId_transaction', 'VisitModeId_x'], errors='ignore', inplace=True)

df.rename(columns={'Rating': 'CSAT Score'}, inplace=True)

df.rename(columns={'AttractionType': 'Product_category'}, inplace=True)

missing_cols = [
    'channel_name',
    'category',
    'Sub-category',
    'Agent Shift',
    'Tenure Bucket',
    'connected_handling_time',
    'Item_price'
]

for col in missing_cols:
    if col not in df.columns:
        df[col] = np.nan

print("Constructed DataFrame 'df' head:")
print(df.head())
print(f"Constructed DataFrame 'df' shape: {df.shape[0]} rows, {df.shape[1]} columns")
````

## 5. Feature Engineering & Preprocessing

````python
data = df.copy()

data["VisitPeriod"] = (
    data["VisitYear"].astype(int).astype(str)
    + "-"
    + data["VisitMonth"].astype(int).astype(str).str.zfill(2)
)

data["VisitPeriod"] = pd.PeriodIndex(data["VisitPeriod"], freq="M")

data = data.sort_values(["VisitPeriod", "UserId", "AttractionId"]).reset_index(drop=True)

print("Dataset prepared for feature engineering.")
print(f"Rows: {len(data):,}")
print(f"Columns: {data.shape[1]}")
````

````python
user_history = (
    data.groupby(["UserId", "VisitPeriod"], sort=True)
    .agg(
        user_total_visits=("CSAT Score", "count"),
        user_avg_rating=("CSAT Score", "mean"),
        user_rating_std=("CSAT Score", "std"),
        user_distinct_attractions=("AttractionId", "nunique"),
    )
    .groupby(level=0)
    .shift(1)
)

user_history = user_history.reset_index()

data = data.merge(
    user_history,
    on=["UserId", "VisitPeriod"],
    how="left"
)

data["user_total_visits"] = data["user_total_visits"].fillna(0)
data["user_avg_rating"] = data["user_avg_rating"].fillna(3.0)
data["user_rating_std"] = data["user_rating_std"].fillna(0)
data["user_distinct_attractions"] = data["user_distinct_attractions"].fillna(0)

data["has_user_history"] = (data["user_total_visits"] > 0).astype(int)

print("User historical features created.")
data[
    [
        "UserId",
        "VisitPeriod",
        "user_total_visits",
        "user_avg_rating",
        "user_distinct_attractions",
        "has_user_history",
    ]
].head()
````

````python
attraction_history = (
    data.groupby(["AttractionId", "VisitPeriod"], sort=True)
    .agg(
        attraction_total_visits=("CSAT Score", "count"),
        attraction_avg_rating=("CSAT Score", "mean"),
        attraction_rating_std=("CSAT Score", "std"),
        attraction_distinct_users=("UserId", "nunique"),
    )
    .groupby(level=0)
    .shift(1)
)

attraction_history = attraction_history.reset_index()

data = data.merge(
    attraction_history,
    on=["AttractionId", "VisitPeriod"],
    how="left"
)

data["attraction_total_visits"] = data["attraction_total_visits"].fillna(0)
data["attraction_avg_rating"] = data["attraction_avg_rating"].fillna(3.0)
data["attraction_rating_std"] = data["attraction_rating_std"].fillna(0)
data["attraction_distinct_users"] = data["attraction_distinct_users"].fillna(0)

data["has_attraction_history"] = (
    data["attraction_total_visits"] > 0
).astype(int)

print("Attraction historical features created.")
data[
    [
        "AttractionId",
        "VisitPeriod",
        "attraction_total_visits",
        "attraction_avg_rating",
        "attraction_distinct_users",
        "has_attraction_history",
    ]
].head()
````

````python
numeric_features = [
    "VisitYear",
    "VisitMonth",
    "user_total_visits",
    "user_avg_rating",
    "user_rating_std",
    "user_distinct_attractions",
    "has_user_history",
    "attraction_total_visits",
    "attraction_avg_rating",
    "attraction_rating_std",
    "attraction_distinct_users",
    "has_attraction_history",
]

categorical_features = [
    "UserContinent",
    "UserRegion",
    "AttractionTypeName",
    "AttractionCityName",
]

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

print(f"\nTotal model features: {len(numeric_features) + len(categorical_features)}")
````

Historical user and attraction statistics were created using only information available before each visit period. Numerical features were prepared for scaling, while categorical features were prepared for encoding. Cold-start cases were handled using neutral/default historical values. This approach reduces data leakage and makes the features suitable for prediction on future visits.
